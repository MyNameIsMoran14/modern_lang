## 1. С помощью регулярных выражений уберите все символы пунктуации из предложения: sentence = "A, very very; irregular_sentence"

```python
import re
sentence = "A, very very; irregular_sentence"
result = re.sub(r"[^A-Za-zА-Яа-яЁё\s]", " ", sentence)
print(result)
```

## 2.	Дана последовательность строк. Выведите строки, содержащие слово, состоящее из двух одинаковых частей (тандемный повтор).
blabla is a tandem repetition  
123123 is good too  
go go  
aaa  


```python
strings = [
    "blabla is a tandem repetition",
    "123123 is good too",
    "go go",
    "aaa"
]

pattern = r"\b(\w+?)\1\b"

for s in strings:
    if re.search(pattern, s):
        print(s)
```

## 3. Дана последовательность строк. Выведите строки, содержащие "cat" в качестве подстроки хотя бы два раза.
catcat  
cat and cat  
catac  
cat  
ccaatt  

```python
strings = [
    "catcat",
    "cat and cat",
    "catac",
    "cat",
    "ccaatt"
]

pattern = r"(?:cat).*cat"

for s in strings:
    if re.search(pattern, s):
        print(s)
```

## 4.	Дана последовательность строк. Выведите строки, содержащие "cat" в качестве слова.
cat

catcat  
concat  
Cat  
"cat"  
!cat?  

```python
strings = [
    "cat",
    "catapult and cat",
    "catcat",
    "concat",
    "Cat",
    '"cat"',
    "!cat?"
]

pattern = r"\bcat\b"

for s in strings:
    if re.search(pattern, s):
        print(s)
```

## 5. Дана последовательность строк. Выведите строки, содержащие обратный слеш "\".

## 6. Дана последовательность строк. В каждой строке замените все вхождения подстроки "human" на подстроку "computer" и выведите полученные строки.
I need to understand the human mind  
humanity  

```python
strings = [
    "I need to understand the human mind",
    "humanity"
]

for s in strings:
    result = re.sub(r"human", "computer", s)
    print(result)
```

## 7. Дана последовательность строк. В каждой строке замените первое вхождение слова, состоящего только из латинских букв "a" (регистр не важен), на слово "argh".
There’ll be no more "Aaaaaaaaaaaaaaa"  
AaAaAaA AaAaAaA  


```python
strings = [
    'There’ll be no more "Aaaaaaaaaaaaaaa"',
    "AaAaAaA AaAaAaA"
]

pattern = r"\b[aA]+\b"

for s in strings:
    result = re.sub(pattern, "argh", s, count=1)
    print(result)
```

## 8.	Дана последовательность строк. В каждой строке поменяйте местами две первых буквы в каждом слове, состоящем хотя бы из двух букв. Буквой считается символ из группы \w.
this is a text  
"this' !is. ?n1ce, 


```python
strings = [
    "this is a text",
    '"this\' !is. ?n1ce,'
]

pattern = r"\b(\w)(\w)(\w*)\b"

for s in strings:
    result = re.sub(pattern, r"\2\1\3", s)
    print(result)
```

## 9.	Дана последовательность строк. В каждой строке замените все вхождения нескольких одинаковых букв на одну букву. Буквой считается символ из группы \w.
attraction  
buzzzz  

```python
strings = [
    "attraction",
    "buzzzz"
]

pattern = r"(\w)\1+"

for s in strings:
    result = re.sub(pattern, r"\1", s)
    print(result)
```

## 10.	Извлеките никнейм пользователя, имя домена и суффикс из данных email адресов.
emails = """zuck26@facebook.com  
page33@google.com    
jeff42@amazon.com"""  


```python
emails = """zuck26@facebook.com
page33@google.com
jeff42@amazon.com"""

pattern = r"([\w.-]+)@([\w.-]+)\.([A-Za-z]+)"

result = re.findall(pattern, emails)

print(result)
```

## 11. С помощью регулярных выражений извлеките ФИО, Email и номера телефонов из предложения.
'''студент Петров В.П. - группа УМР11 тел. +7(918)240-04-40  
Email: petrov@mail.ru доп.тел +7(926)718-33-22'''

```python
text = """студент Петров В.П. - группа УМР11 тел. +7(918)240-04-40 Email: petrov@mail.ru доп.тел +7(926)718-33-22""" 

fio = re.findall(r"[А-ЯЁ][а-яё]+\s+[А-ЯЁ]\.[А-ЯЁ]\.", text) 
email = re.findall(r"[\w.-]+@[\w.-]+\.[A-Za-z]+", text)
phones = re.findall(r"\+7\(\d{3}\)\d{3}-\d{2}-\d{2}", text) 
print("ФИО:", fio) 
print("Email:", email) 
print("Телефоны:", phones)
```
